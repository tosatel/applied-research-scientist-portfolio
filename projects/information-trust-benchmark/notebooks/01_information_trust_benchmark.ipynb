# Information Trust & Misinformation Detection Benchmark

## Research Question

Can multidimensional information-trust signals improve misinformation detection, calibration, and interpretability compared with content-only and source-only baselines?

**Dataset:** Synthetic information-trust signals.

**Important:** These results are methodological demonstrations, not validated real-world misinformation detection.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "src" / "experiment.py").exists():
    ROOT = ROOT.parent

assert (ROOT / "src" / "experiment.py").exists(), "Open this notebook from the project or notebooks directory."

results = pd.read_csv(ROOT / "results" / "sample_results.csv")
results
```

## Model Comparison

We compare content-only, source-only, and multidimensional models using classification performance and probability-quality metrics.


```python
metrics = ["precision", "recall", "f1", "roc_auc"]
display(results.set_index("model")[metrics].round(3))

results.plot(
    x="model",
    y=["f1", "roc_auc"],
    kind="bar",
    figsize=(9, 5),
    rot=0,
    title="Information Trust Benchmark — Model Comparison",
)
plt.ylabel("Score")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```

## Probability Quality

Lower Brier score and log loss indicate better probabilistic predictions on this dataset. These metrics alone do not establish calibration across probability bins.


```python
results.plot(
    x="model",
    y=["brier_score", "log_loss"],
    kind="bar",
    figsize=(9, 5),
    rot=0,
    title="Probability Quality Comparison",
)
plt.ylabel("Loss (lower is better)")
plt.tight_layout()
plt.show()
```

## Model Interpretability

Standardized logistic-regression coefficients describe associations between features and model predictions. They should not be interpreted as causal effects.


```python
coefficients = pd.read_csv(ROOT / "results" / "coefficients.csv")
combined = coefficients[
    coefficients["model"] == "Multidimensional"
].sort_values("standardized_coefficient")

display(combined)

combined.plot(
    x="feature",
    y="standardized_coefficient",
    kind="barh",
    legend=False,
    figsize=(9, 5),
    title="Multidimensional Model — Standardized Coefficients",
)
plt.xlabel("Standardized coefficient")
plt.tight_layout()
plt.show()
```

## Limitations and Future Work

- Synthetic data may encode relationships that are simpler than real information environments.
- Results from one train/test split do not establish statistical significance.
- Source credibility is not equivalent to factual truth.
- Strong predictive performance does not establish causal validity.
- Real-world validation, repeated trials, calibration curves, and distribution-shift experiments are needed.
