# LLM Evaluation Lab

Reproducible starter notebook for evaluating LLM outputs.


```python
from pathlib import Path
import sys, pandas as pd
PROJECT = Path.cwd()
if PROJECT.name == 'notebooks': PROJECT = PROJECT.parent
sys.path.insert(0, str(PROJECT / 'src'))
from evaluate import read_csv, evaluate, summarize
```

## Load benchmark and model outputs


```python
benchmark = read_csv(PROJECT / 'data' / 'benchmark.csv')
predictions = read_csv(PROJECT / 'data' / 'sample_model_outputs.csv')
pd.DataFrame(benchmark).head()
```

## Evaluate responses


```python
results = evaluate(benchmark, predictions)
results_df = pd.DataFrame(results)
results_df.head()
```

## Aggregate by model and condition


```python
summary_df = pd.DataFrame(summarize(results))
summary_df
```
